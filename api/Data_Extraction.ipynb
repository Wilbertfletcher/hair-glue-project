```python
import os
import time
from chemspipy import ChemSpider

def main():
    # 1. Initialize the ChemSpider client
    api_key = os.environ.get("CHEMSPIDER_API_KEY")
    if not api_key:
        raise SystemExit("Set CHEMSPIDER_API_KEY environment variable before running.")
    cs = ChemSpider(api_key)

    try:
        print("--- 1. Simple Search: Searching for 'Aspirin' ---")
        # Perform an asynchronous search for the compound name
        results = cs.search('Aspirin')
        
        # Wait for the background search task to complete
        results.wait()

        if results.count > 0:
            # Get the top matched compound object
            top_result = results[0]
            
            print(f"Total results found: {results.count}")
            print(f"Top Match CSID: {top_result.record_id}")
            print(f"Common Name: {top_result.common_name}")
            print(f"Molecular Formula: {top_result.molecular_formula}")
            print(f"SMILES String: {top_result.smiles}")
            print(f"Molecular Weight: {top_result.molecular_weight}")

            print("\n--- 2. Download 2D Structure Image ---")
            # The image property returns bytes representing a PNG file
            image_filename = f"compound_{top_result.record_id}_2d.png"
            with open(image_filename, 'wb') as f:
                f.write(top_result.image)
            print(f"Successfully saved 2D structure image to '{image_filename}'")

        else:
            print("No results found for the given query.")

        print("\n--- 3. Advanced Filtering: Search by Molecular Formula ---")
        # Search for compounds with the exact formula 'C9H8O4'
        target_formula = 'C9H8O4'
        print(f"Initiating filter for formula: {target_formula}")
        
        # This returns a query_id which is used to fetch the actual results
        query_id = cs.filter_formula(target_formula)
        
        # Adding a slight delay to ensure the server processes the filter request
        time.sleep(1)
        
        # Retrieve the first 3 record IDs matching the formula
        matched_record_ids = cs.filter_results(query_id, start=0, count=3)
        print(f"Top 3 CSIDs matching {target_formula}: {matched_record_ids}")

        if matched_record_ids:
            print("\n--- 4. Batch Detail Retrieval ---")
            # Fetch detailed dictionaries for the list of IDs efficiently
            batch_details = cs.get_details_batch(matched_record_ids)
            for detail in batch_details:
                name = detail.get('CommonName', 'N/A')
                mass = detail.get('AverageMass', 'N/A')
                print(f" - CSID {detail.get('Id')}: {name} (Avg Mass: {mass})")

    except Exception as e:
        print(f"Error: {e}")

main()
```
